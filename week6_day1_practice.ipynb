## Task 4 — Build a missing-value report

````python
# Your code here
report = pd.DataFrame({
    'missing_count' : penguins.isnull().sum(),
    'missing_percent' : (penguins.isnull().mean()* 100).round(1)
})
report = report[report['missing_count'] > 0]

report.head()
````

## Task 8 — Fill the numeric columns

In `clean_penguins`, fill the missing values in all four measurement columns using the
**median** of each column:

- `bill_length_mm`
- `bill_depth_mm`
- `flipper_length_mm`
- `body_mass_g`

Then confirm each one has zero missing values.

````python
# Your code here
clean_penguins['bill_length_mm'] = clean_penguins['bill_length_mm'].fillna(clean_penguins['bill_length_mm'].median)
clean_penguins['bill_depth_mm'] = clean_penguins['bill_depth_mm'].fillna(clean_penguins['bill_depth_mm'].median)
clean_penguins['flipper_length_mm'] = clean_penguins['flipper_length_mm'].fillna(clean_penguins['flipper_length_mm'].median)
clean_penguins['body_mass_g'] = clean_penguins['body_mass_g'].fillna(clean_penguins['body_mass_g'].median)


print("Missing after bill_length_mm: ", clean_penguins['bill_length_mm'].isnull().sum())
print("Missing after bill_depth_mm: ", clean_penguins['bill_depth_mm'].isnull().sum())
print("Missing after flipper_length_mm: ", clean_penguins['flipper_length_mm'].isnull().sum())
print("Missing after body_mass_g: ", clean_penguins['body_mass_g'].isnull().sum())
````

## Task 10 — Try 'Unknown' instead

Filling `sex` with the mode is a real decision with consequences — it invents a sex for
11 penguins that nobody recorded.

Make a *second* copy of the original data, and this time fill `sex` with the string
`"Unknown"` instead. Print the value counts for both versions side by side.

Which do you prefer, and why?

````python
# Your code here
trial = penguins.copy()
````

````python
trial['sex'] = trial['sex'].fillna('Unknown')

before = pd.DataFrame({
    'before_missing': penguins.isnull().sum()
})
after = pd.DataFrame({
    'after_missing': trial.isnull().sum()
})

comparison = before.join(after)
comparison
````
